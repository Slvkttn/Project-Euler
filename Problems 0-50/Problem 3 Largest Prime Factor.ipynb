# Problem 3 - Largest Prime Factor
The prime factors of 13195 are `5, 7, 13 and 29`

What is the largest prime factor of the number `600851475143`?

## What is a prime factor?

A prime factor is a prime number that divides a given integer exactly, without leaving a remainder.

### Tree Factor Method

A common way to find prime factors is through the Factor Tree Method:

1. **Start**: Take the given number, for example, 20
2. **Split**: Find any two factors that multiply to make 20 (e.g. 4 x 5)
3. **Identify Primes**: If a factor is a prime, circle it.
4. **Repeat**: If a factor is not a prime, split it further (e.g. 4 becomes 2 x 2) until all factors are prime.
5. **Result**: The circled numbers (`2 x 2 x 5`) are the prime factors of 20

### Trial Division

Another way is the trial division method, which is the method I will implement programmatically here.

1. **Divide by 2**: If the number is even, divide by 2 repeatedly until it is odd.
2. **Divide by Odd Numbers**: Divide the result by odd numbers `(3, 5, 7, ..)` starting from 3 up to the square root of the remaining number `sqrt(n)`
3. **Repeat**: For every divisor `i`, divide the quotient by `i` as long as it is divisible.
4. **Final Prime**: If the remaining quotient is greater than 1, then that remaining quotient is also a prime factor.

### Packages

```python
import math
```

### Solution Function

```python
def largest_prime_factor(num: int) --> list[int]:
    """
    Calculates all prime factors of a given integer using trial division.
    To find the largest prime factor, one simply collects the number at the index [-1]

    Args:
      num (int): The integer to factorize into its prime factors.

    Returns:
      list[int]: A list of all prime factors in ascending order.
    """

    print(f"Finding Prime Factors for {num} ...")
    prime_factors = []

    twos = 0
    
    while num % 2 == 0:
        num //= 2
        twos += 1

    prime_factors.extend([2] * twos)

    if num == 1:
        print("Looks like you passed a basic multiple of two. Who would have seen that coming?")
        print("---------------------------------------------------")
        return prime_factors

    print("---------------------------------------------------")
    print(f"After dividing by two {twos} times, the remainder is {num}")
    print(f"However we only have to check odd numbers up to sqrt(n) (e.g. while divisor * divisor <= remainder)")
    print(f"We are checking for prime factors up to {int(math.ceil(math.sqrt(num)))}")
    print("---------------------------------------------------")
    
    divisor = 3
    
    while divisor**2 <= num:
        while num % divisor == 0:
            print(f"Looks like {divisor} is prime.")
            prime_factors.append(divisor)
            num //= divisor
        divisor += 2

    if num > 1:
        print(f"Looks like {num} is the final prime.")
        print("---------------------------------------------------")
        prime_factors.append(num)

    return prime_factors
```

### Solution to Problem

```python
%%time
f"The solution is {largest_prime_factor(600851475143)[-1]}"
```
