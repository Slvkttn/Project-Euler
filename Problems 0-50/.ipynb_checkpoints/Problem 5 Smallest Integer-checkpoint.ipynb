## Smallest Multiple


This problem is best solved by considering the **Least Common Multiple (LCM)**  

The relationship between the LCM and the **Greatest Common Divisor (GCD)** for any two numbers is given by:  

$\text{LCM}(a, b) = \frac{a \times b}{\text{GCD}(a, b)}$

Luckily this is a part of the default `math` package via the `.lcm` method. We will simply use this in a basic function call to get our solution.

```python
import math
```

```python
def evenly_divisible(limit: int = 20) -> int:
    """
    Calculates the smallest integer evenly divisible by all numbers from 1 to the input limit.
    
    Args:
        limit (integer, optional): Upper bound of sequence of integers to check. Defaults to 20 (Project Euler Problem does 20)

    Returns:
        int: The smallest positive integer that is evenly divisible by all numbers from 1 to `limit`
        
    """
    print(f"Calculating LCM for integers between 1 and {limit}...")

    return math.lcm(*range(1, limit + 1))
```

## Solution

```python
%%time
print(evenly_divisible())
```
