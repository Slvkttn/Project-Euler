## Special Pythagorean Triplet
A Pythagorean triplet is a set of three natural numbers, $a < b < c$, for which $a^2 + b^2 = c^2$

For example: $3^2 + 4^2 = 9 + 16 = 25 = 5^2$  

There exists exactly one Pythagorean triplet for which $a + b + c = 1000$  
Find the product abc.  

---

### Solution
This problem can be solved using some basic algebra, which is demonstrated in the following...

**The first constraint is that:** $a^2 + b^2 = c^2$

**We also know that:** $a + b + c = s$ meaning that $c = s - a- b$  

**We can substitute this equation into the first:**  
--> $a^2 + b^2 = (s - a - b)^2$  

**Expanding the right hand side:**  
--> $\large^2 + b^2 = s^2 + a^2 + b^2 - 2sa - 2sb + 2ab$

**Subtracting $a^2 + b^2$ from both sides:**  
--> $\large0 = s^2 - 2sa - 2sb + 2ab$

**Solve for b:**  
--> $\large 2sb - 2ab = s^2 - 2sa$  
--> $\large b(2s - 2a) = s^2 - 2sa$  
  
--> $\large b = \frac{s^2 - 2sa}{2(s - a)}$

```python
def get_pythagorean_triplet_product(target_sum: int = 1000) -> str:
    """
    Finds the product of the Pythagorean triplet whose sum is `target_sum`.

    This algorithm uses algebra to solve the problem in O(n) time
    and O(1) space. 
    
    By substituting c = target_sum - a - b into the Pythagorean
    theorem (a^2 + b^2 = c^2), we can isolate b as a function of a and the sum:
    b = (target_sum^2 - 2 * target_sum * a) / (2 * (target_sum - a))

    More documentation can be found in my Jupyter Notebook.

    Args:
        target_sum: The desired sum of the three natural numbers a, b, and c.

    Returns:
        A formatted string displaying the triplet and its product in the format
        "Triplet: a x b x c = product". Returns "No triplet found" if none exists.
    """
    # Because a < b < c ... Then a must be strictly less than a third of the target sum.
    max_a = target_sum // 3

    # Loop from a to max_a
    for a in range(1, max_a):
        numerator = (target_sum * target_sum) - (2 * target_sum * a)
        denominator = 2 * (target_sum - a)

        # Check if b evaluates to a whole integer
        if numerator % denominator == 0:
            b = numerator // denominator
            c = target_sum - a - b

            # Ensure the problem's strict inequality constraints are met
            if a < b < c:
                return f"Triplet: {a} x {b} x {c} = {a * b * c}"

    # If no triplet exists for target_sum return default message
    return "No triplet found"
```

---

### Answer

This leaves us with a function that iterate through possible values of `a` in `O(n)` time and calculates b in `O(1)` space. 

```python
%%time
get_pythagorean_triplet_product(1000)
```
