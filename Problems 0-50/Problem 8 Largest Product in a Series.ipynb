Find the thirteen adjacent digits in the 1000-digit number that have the greatest product. What is the value of this product?

### Solution

```python
import math
import random

random.seed(141414) # For reproducibility of second additional test
```

```python
def find_max_adjacent_product(digit_str: str, n: int) -> tuple[int, str]:
    """
    Finds the maximum product of n adjacent digits in a string.

    Optimizes the search by ignoring zeros and using a sliding window with O(1)
    product updates.

    Args:
        digit_str: The input string of numbers.
        n: The number of adjacent digits to multiply.

    Returns:
        A tuple containing (product, sequence). Returns (None, "") if n is too
        large, or (0, "") if no product above 0 is found.
    """

    # Safeguarding against input string of numbers with whitespaces
    s = "".join(digit_str.split())

    # If user inputs adjacent digits larger than the input string length, immediately halt
    if len(s) < n:
        print(
            f"Error: The requested digit count ({n}) is larger than the total length of the input string ({len(s)} digits)."
        )
        return 0, ""

    # Placeholder max product and string of adjacent digits
    max_p, best = -1, ""

    # We can split at 0s as multiplying with 0 always yields 0
    for chunk in s.split("0"):
        if len(chunk) < n:
            continue

        # Convert string chunks into lists of integers so we can do .prod() on it
        nums, p = [int(d) for d in chunk], math.prod([int(d) for d in chunk[:n]])

        # Sliding window over the indices of the list of integers
        for i in range(n - 1, len(nums)):
            if i >= n:

                # Calculate the product of the current window
                p = (p // nums[i - n]) * nums[i]
                
            if p > max_p:

                # Update the max product and the string of adjacent digits producing this product
                max_p, best = p, chunk[i - n + 1 : i + 1]


    # Handling case where no product over 0 is found
    if max_p == -1:
        print(
            f"No product was found over 0 for {n} adjacent digits in the input string."
        )
        return 0, ""

    # Return maximum product and the string of adjacent digits that produced it
    return max_p, best
```

```python
%%time
product, sequence = find_max_adjacent_product(input_numbers, n_digits)

equation = " x ".join(list(sequence))

print(f"Greatest product of {n_digits} adjacent digits: {equation} = {product:,}")
```

**Let us try to generate a massive amount of random integers, for instance 10 million, and then look for 123 adjacent digits instead**

```python
total_digits = 10000000
n_digits_to_find = 123

large_number_string = "".join(random.choices("0123456789", k=total_digits))
```

```python
%%time

product, sequence = find_max_adjacent_product(large_number_string, n_digits_to_find)

equation = " x ".join(list(sequence))
print(f"Greatest product of {n_digits_to_find} adjacent digits: {equation} = {product:,}")
```

**We see that the implementation works fast with extremely large values and computes essentially in O(L) where L is the length of the input string**
