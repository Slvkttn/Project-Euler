## Largest Palindrome Product

A palindromic number reads the same both ways. The largest palindrome made from the product of two 2-digit numbers is `9009 = 91 x 99`

Find the largest palindrome made from the product of two 3-digit numbers.

```python
def largest_palindrome_product(digits: int = 3) -> int:
    """
    Finds the largest palindrome made from the product of two n-digit numbers.

    Args:
      digits (int): The number of digits for the product

    Returns:
      int: The largest palindrome product found
    """

    if not isinstance(digits, int):
        raise TypeError(f"Input must be an integer. You tried to pass {type(digits).__name__}")
    
    upper_bound = 10**digits - 1 # Upper product term
    lower_bound = 10**(digits - 1) # Lower product term
    
    max_palindrome = 0
    max_term_one = max_term_two = 0

    # Outer loop: Start at the upper bound and loop down by 1 until the lower bound
    for first_product in range(upper_bound, lower_bound, -1):
       
        # If the first product times the upper bound is smaller than the current largest palindrome
        # Then no smaller first product can possibly yield a new larger palindrome.
        if first_product * upper_bound <= max_palindrome:
            break

        # Inner Loop: Start at the upper_bound, finish at the first_product
        for second_product in range(upper_bound, first_product, -1):
            
            product = first_product * second_product

            # If the product is smaller than the current max, move on.
            if product <= max_palindrome:
                break
            
            # Check if number is actually a palindrome and store as current max
            s = str(product)
            if s == s[::-1]:
                max_palindrome = product
                max_term_one, max_term_two = first_product, second_product

    print(f"For {digits}-digit numbers...")
    print(f"The largest palindrome product is {max_term_one} x {max_term_two} = {max_palindrome}")
    return max_palindrome
```
